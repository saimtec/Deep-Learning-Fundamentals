# Recurrent Neural Network (RNN)

A **Recurrent Neural Network (RNN)** is a neural network designed for **sequence data**, where the order of the data matters.

Examples of sequence data:

* Text
* Speech
* Time series
* Stock/forex prices
* Sensor readings
* User activity sequences

Unlike a normal neural network, an RNN maintains information from previous time steps using a **hidden state**.

---

# 1. Why Do We Need RNNs?

Consider the sentence:

> "I live in Lahore and I speak ____."

To predict the next word correctly, the model needs information from the previous words.

The words are not independent.

For example:

```text
I → live → in → Lahore → and → I → speak → ...
```

The order matters.

A normal feed-forward neural network does not naturally maintain information about previous inputs.

RNNs solve this by passing a **hidden state** from one time step to the next.

---

# 2. ANN vs CNN vs RNN

| Model | Main Purpose        | Typical Data              |
| ----- | ------------------- | ------------------------- |
| ANN   | General prediction  | Tabular data              |
| CNN   | Spatial patterns    | Images                    |
| RNN   | Sequential patterns | Text, time series, speech |

### ANN

```text
Input → Hidden Layers → Output
```

Each input is generally processed without a recurrent memory mechanism.

### CNN

```text
Image
  ↓
Convolution
  ↓
Feature Maps
  ↓
Pooling
  ↓
Prediction
```

CNNs are particularly useful for spatial relationships.

### RNN

```text
x₁ → RNN → h₁
          ↓
x₂ → RNN → h₂
          ↓
x₃ → RNN → h₃
          ↓
...
```

The hidden state carries information forward.

---

# 3. Basic RNN Architecture

An RNN processes a sequence one element at a time.

For a sequence:

```text
x₁, x₂, x₃, x₄
```

the RNN processes:

```text
x₁ → h₁
      ↓
x₂ → h₂
      ↓
x₃ → h₃
      ↓
x₄ → h₄
```

Each hidden state contains information about the current input and previous information.

A useful conceptual diagram is:

```text
                  ┌───────────────┐
x₁ ──────────────►│               │──► h₁
                  │     RNN       │
                  └───────────────┘
                         │
                         ▼
                  ┌───────────────┐
x₂ ──────────────►│     RNN       │──► h₂
                  └───────────────┘
                         │
                         ▼
                  ┌───────────────┐
x₃ ──────────────►│     RNN       │──► h₃
                  └───────────────┘
```

# 4. RNN Hidden State

The most important idea in an RNN is the **hidden state**.

At time step $t$:

$$
h_t = \tanh(W_{xh}x_t + W_{hh}h_{t-1} + b_h)
$$

The new hidden state depends on:

* Current input $x_t$
* Previous hidden state $h_{t-1}$
* Input-to-hidden weights $W_{xh}$
* Hidden-to-hidden weights $W_{hh}$
* Bias $b_h$

The hidden state is then passed to the next time step.

---

# 5. RNN Output

The output can be calculated using:

$$
y_t = W_{hy}h_t + b_y
$$

For classification, an activation such as Softmax can be applied.

For example:

```text
Hidden State
     ↓
Dense Layer
     ↓
Softmax
     ↓
Class Probabilities
```

---

# 6. Where Do the Parameters Come From?

### $x_t$

Input at the current time step.

### $h_t$

Current hidden state.

### $h_{t-1}$

Previous hidden state.

### $W_{xh}$

Weights connecting input to hidden state.

### $W_{hh}$

Weights connecting previous hidden state to current hidden state.

### $W_{hy}$

Weights connecting hidden state to output.

### $b_h$

Hidden-state bias.

### $b_y$

Output bias.

---

# 7. Why Are RNN Weights Reused?

The same RNN parameters are used at every time step.

For example:

```text
x₁ → [same RNN weights] → h₁
x₂ → [same RNN weights] → h₂
x₃ → [same RNN weights] → h₃
```

The model does not create a completely new set of weights for every word/time step.

This greatly reduces the number of parameters.

---

# 8. Unrolling an RNN

An RNN can be represented as a loop:

```text
          ┌───────────────┐
          │     RNN       │
          └───────┬───────┘
                  │
                  ▼
              Hidden State
                  │
                  └──────────┐
                             │
Next input ─────────────────►│
```

When we "unroll" it:

```text
x₁ → RNN → h₁ → RNN → h₂ → RNN → h₃
```

This makes the sequence processing easier to visualize.

---

# 9. Many-to-One RNN

Used when a complete sequence produces one output.

Example:

```text
"I love this movie"
        ↓
       RNN
        ↓
     Positive
```

Applications:

* Sentiment classification
* Sequence classification

---

# 10. Many-to-Many RNN

Each time step can produce an output.

```text
x₁ → y₁
x₂ → y₂
x₃ → y₃
```

Applications include:

* Sequence labeling
* Some speech tasks
* Token-level prediction

---

# 11. Many-to-One vs Many-to-Many

| Architecture | Input    | Output           |
| ------------ | -------- | ---------------- |
| Many-to-One  | Sequence | One output       |
| Many-to-Many | Sequence | Multiple outputs |

# 12. RNN Training

RNNs are trained using **Backpropagation Through Time (BPTT)**.

The RNN is unfolded across time and gradients are propagated backward through the sequence.

Conceptually:

```text
Forward:

x₁ → h₁ → h₂ → h₃ → output

Backward:

gradient ← gradient ← gradient ← gradient
```

---

# 13. Vanishing Gradient Problem

One major limitation of traditional RNNs is the **vanishing gradient problem**.

During BPTT, gradients can involve repeated multiplication through many time steps.

If these values become very small:

$$
0.5 \times 0.5 \times 0.5 \times \cdots
$$

the gradient can become extremely small.

Eventually, earlier time steps receive almost no useful learning signal.

---

# 14. Why Is Vanishing Gradient a Problem?

Suppose:

```text
The movie was extremely long, but the ending was amazing.
```

To understand "amazing", the model may need information from much earlier in the sequence.

If the sequence is long, a basic RNN can struggle to preserve useful information across many time steps.

This is one reason LSTM and GRU architectures were developed.

---

# 15. Advantages of RNN

* Designed for sequential data
* Maintains hidden state
* Can process variable-length sequences
* Uses shared parameters across time steps
* Simpler than LSTM and GRU
* Useful for learning basic sequence concepts

---

# 16. Limitations of RNN

* Vanishing gradient
* Difficulty learning long-term dependencies
* Sequential computation can limit parallelization
* Basic RNNs can struggle with long sequences

---

# 17. RNN in Keras

A typical architecture is:

```text
Input
  ↓
Embedding
  ↓
SimpleRNN
  ↓
Dense
  ↓
Softmax
```

For numerical sequence prediction, an embedding layer may not be necessary.


````python
import numpy as np
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense


# ============================================================
# 1. CREATE TRAINING DATA
# ============================================================

X = np.array([
    [1, 2, 3],
    [2, 3, 4],
    [3, 4, 5],
    [4, 5, 6],
    [5, 6, 7],
    [6, 7, 8],
    [7, 8, 9]
], dtype=np.float32)

y = np.array([
    4,
    5,
    6,
    7,
    8,
    9,
    10
], dtype=np.float32)


# ============================================================
# 2. RESHAPE DATA FOR RNN
# ============================================================

X = X.reshape(
    X.shape[0],
    X.shape[1],
    1
)

print("X shape:", X.shape)
print("y shape:", y.shape)


# ============================================================
# 3. CREATE RNN MODEL
# ============================================================

model = Sequential([
    SimpleRNN(
        16,
        activation="tanh",
        input_shape=(3, 1)
    ),

    Dense(1)
])


# ============================================================
# 4. COMPILE MODEL
# ============================================================

model.compile(
    optimizer="adam",
    loss="mse"
)


# ============================================================
# 5. DISPLAY MODEL
# ============================================================

model.summary()


# ============================================================
# 6. TRAIN MODEL
# ============================================================

model.fit(
    X,
    y,
    epochs=200,
    verbose=0
)


# ============================================================
# 7. MAKE A PREDICTION
# ============================================================

test_sequence = np.array(
    [8, 9, 10],
    dtype=np.float32
)

test_sequence = test_sequence.reshape(
    1,
    3,
    1
)


prediction = model.predict(
    test_sequence,
    verbose=0
)


print("\nInput sequence:")
print([8, 9, 10])

print("\nPredicted next value:")
print(prediction[0][0])

print("\nExpected value:")
print(11)
````
