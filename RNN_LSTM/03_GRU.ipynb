# Gated Recurrent Unit (GRU)

A **GRU (Gated Recurrent Unit)** is another gated recurrent architecture designed to handle sequence data and long-term dependencies.

GRU is similar to LSTM but has a **simpler architecture**.

---

# 1. Why GRU?

Basic RNNs can struggle with long-term dependencies.

LSTM solves this using:

* Cell state
* Hidden state
* Forget gate
* Input gate
* Output gate

GRU simplifies this architecture.

A GRU uses two main gates:

1. Update Gate
2. Reset Gate

---

# 2. GRU Architecture

Conceptually:

```text
Previous Hidden State
         │
         ├─────────────┐
         │             │
         ▼             ▼
   Reset Gate      Update Gate
         │             │
         ▼             │
 Candidate State        │
         │             │
         └──────┬──────┘
                ▼
         New Hidden State
```

# 3. GRU State

Unlike LSTM, GRU does not maintain a separate cell state.

It uses:

$$
h_t
$$

as its main state.

Therefore:

```text
LSTM:
Hidden State + Cell State

GRU:
Hidden State
```

---

# 4. Update Gate

The update gate determines how much previous information should be retained.

A common formulation is:

$$
z_t =
\sigma(W_z[x_t,h_{t-1}]+b_z)
$$

The value is between 0 and 1.

Conceptually:

```text
0 → rely more on new information
1 → retain more previous information
```

The exact interpretation depends on the equation convention being used.

---

# 5. Reset Gate

The reset gate controls how much previous information is used when creating the candidate state.

$$
r_t =
\sigma(W_r[x_t,h_{t-1}]+b_r)
$$

Conceptually:

```text
Reset gate
    ↓
How much previous information should
be considered for the candidate?
```

---

# 6. Candidate Hidden State

A candidate hidden state can be calculated as:

$$
\tilde{h}_t =
\tanh(W_h[x_t,r_t \odot h_{t-1}]+b_h)
$$

The reset gate controls the contribution of the previous hidden state.

---

# 7. New Hidden State

A common GRU formulation is:

$$
h_t =
(1-z_t)\odot h_{t-1}
+
z_t\odot\tilde{h}_t
$$

This combines:

* Previous hidden state
* Candidate hidden state

according to the update gate.

---

# 8. Complete GRU Flow

```text
                 xₜ
                  │
        ┌─────────┴─────────┐
        │                   │
        ▼                   ▼
   Reset Gate          Update Gate
        │                   │
        ▼                   │
Previous hₜ₋₁               │
        │                   │
        ▼                   │
Candidate h̃ₜ               │
        │                   │
        └─────────┬─────────┘
                  ▼
               hₜ
```

---

# 9. GRU vs LSTM

| Feature             | LSTM         | GRU     |
| ------------------- | ------------ | ------- |
| Hidden State        | Yes          | Yes     |
| Separate Cell State | Yes          | No      |
| Forget Gate         | Yes          | No      |
| Input Gate          | Yes          | No      |
| Output Gate         | Yes          | No      |
| Update Gate         | No           | Yes     |
| Reset Gate          | No           | Yes     |
| Architecture        | More complex | Simpler |
| Parameters          | More         | Fewer   |

---

# 10. Why GRU Can Be Faster

GRU has fewer gates and does not maintain a separate cell state.

Therefore, compared with an equivalent LSTM architecture, it can have fewer parameters and a simpler computation.

Conceptually:

```text
LSTM
More gates + cell state
        ↓
More complex

GRU
Fewer gates + hidden state
        ↓
Simpler
```

The actual speed depends on the implementation, hardware, sequence length, and model configuration.

---

# 11. GRU Advantages

* Simpler than LSTM
* Fewer parameters than comparable LSTM configurations
* Can model long-term dependencies
* Often useful when computational efficiency matters
* Good option for sequence modeling

---

# 12. GRU Limitations

* Still recurrent
* Still processes sequences sequentially
* Less explicit memory structure than LSTM
* Performance depends on the specific dataset and task

---

# 14. RNN vs LSTM vs GRU

| Feature                | RNN          | LSTM              | GRU                     |
| ---------------------- | ------------ | ----------------- | ----------------------- |
| Basic recurrent state  | Yes          | Yes               | Yes                     |
| Separate cell state    | No           | Yes               | No                      |
| Gates                  | No           | 3 main gates      | 2 main gates            |
| Long-term dependencies | Difficult    | Better            | Better                  |
| Complexity             | Low          | High              | Medium                  |
| Parameters             | Lowest       | Highest           | Usually fewer than LSTM |
| Main idea              | Hidden state | Controlled memory | Gated hidden state      |
