# Long Short-Term Memory (LSTM)

**LSTM (Long Short-Term Memory)** is a special type of recurrent neural network designed to handle **long-term dependencies** better than a basic RNN.

LSTM was introduced to address problems such as the **vanishing gradient problem** encountered by traditional RNNs.

---

# 1. Why LSTM?

A basic RNN has:

```text
Input
  ↓
Hidden State
  ↓
Next Hidden State
```

The hidden state has to carry information through the sequence.

For long sequences, important information can gradually disappear.

LSTM introduces a dedicated **cell state** and **gates** to control information.

---

# 2. LSTM Architecture

The main components are:

1. Cell State
2. Forget Gate
3. Input Gate
4. Candidate Cell State
5. Output Gate
6. Hidden State

Conceptually:

```text
             Previous Cell State
                    │
                    ▼
             ┌─────────────┐
             │ Forget Gate │
             └──────┬──────┘
                    │
                    ▼
                Remove
                information
                    │
                    +
                    │
             New information
                    │
                    ▼
              Cell State
                    │
                    ▼
              Output Gate
                    │
                    ▼
              Hidden State
```

# 3. Two Types of State

An LSTM maintains two important states.

## Hidden State

$$
h_t
$$

This represents the current output/state of the LSTM.

## Cell State

$$
C_t
$$

This acts as a long-term memory path.

Therefore:

```text
LSTM
 ├── Cell State Cₜ
 └── Hidden State hₜ
```

---

# 4. Forget Gate

The forget gate decides what information should be removed from the previous cell state.

$$
f_t =
\sigma(W_f[h_{t-1},x_t]+b_f)
$$

The sigmoid function produces values between 0 and 1.

```text
0 → Forget
1 → Keep
```

For example:

```text
Previous memory
      ↓
Forget Gate
      ↓
How much should remain?
```

---

# 5. Input Gate

The input gate determines how much new information should be stored.

$$
i_t =
\sigma(W_i[h_{t-1},x_t]+b_i)
$$

It controls the amount of new information entering the cell state.

---

# 6. Candidate Cell State

The LSTM creates candidate information:

$$
\tilde{C}_t =
\tanh(W_C[h_{t-1},x_t]+b_C)
$$

This represents possible new information that can be added to memory.

---

# 7. Updating the Cell State

The new cell state is:

$$
C_t =
f_t \odot C_{t-1}
+
i_t \odot \tilde{C}_t
$$

This equation combines:

* Old memory that should be kept
* New information that should be stored

---

# 8. Output Gate

The output gate controls what information becomes the new hidden state.

$$
o_t =
\sigma(W_o[h_{t-1},x_t]+b_o)
$$

Then:

$$
h_t =
o_t \odot \tanh(C_t)
$$

The hidden state becomes the output passed to the next time step.

---

# 9. Complete LSTM Flow

The overall process is:

```text
Previous hₜ₋₁
Previous Cₜ₋₁
      │
      ├──────────────► Forget Gate
      │
Current xₜ ──────────► Input Gate
      │
      ├──────────────► Candidate Memory
      │
      └──────────────► Output Gate
                         │
                         ▼
                    New Cₜ
                         │
                         ▼
                    New hₜ
```

---

# 10. What Each Gate Does

| Component       | Main Job                               |
| --------------- | -------------------------------------- |
| Forget Gate     | Decides what old information to remove |
| Input Gate      | Decides what new information to store  |
| Candidate State | Creates possible new memory            |
| Cell State      | Carries long-term information          |
| Output Gate     | Decides what to output                 |
| Hidden State    | Current output representation          |

---

# 11. Why Sigmoid?

The sigmoid function is:

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

Its output lies between:

$$
0 \leq \sigma(x) \leq 1
$$

This makes it useful for gates.

For example:

```text
0.0 → almost completely blocked
0.5 → partially allowed
1.0 → almost completely allowed
```

---

# 12. Why Tanh?

The tanh function produces values between:

$$
-1 \leq \tanh(x) \leq 1
$$

It is useful for representing candidate information and transforming the cell state before producing the hidden state.

---

# 13. Why LSTM Handles Long-Term Dependencies Better

The cell state provides a relatively direct path for information to flow through time.

Instead of forcing all information through a single hidden state, LSTM uses controlled memory operations.

Conceptually:

```text
Old Memory
    ↓
Forget ──► Remove unnecessary information
    ↓
Input ───► Add useful information
    ↓
New Memory
    ↓
Output ──► Produce current hidden state
```

---

# 14. LSTM Example

Consider:

> "I grew up in Pakistan. After many years, I still speak ____."

A sequence model may need earlier information to predict:

```text
Urdu
```

An LSTM can preserve useful information through its cell state and gates.

---

# 15. LSTM vs Basic RNN

| Feature                | RNN       | LSTM                           |
| ---------------------- | --------- | ------------------------------ |
| Hidden State           | Yes       | Yes                            |
| Cell State             | No        | Yes                            |
| Gates                  | No        | Yes                            |
| Long-term dependencies | Difficult | Better supported               |
| Architecture           | Simple    | More complex                   |
| Parameters             | Fewer     | More                           |
| Training               | Simpler   | More computationally expensive |

---

# 16. LSTM Advantages

* Better handling of long-term dependencies
* Reduces problems associated with vanishing gradients
* Useful for text and sequence data
* Useful for time-series problems
* Can preserve important information for longer sequences

---

# 17. LSTM Limitations

* More parameters than a basic RNN
* More computationally expensive
* More complex architecture
* Sequential recurrence can still limit parallelization
