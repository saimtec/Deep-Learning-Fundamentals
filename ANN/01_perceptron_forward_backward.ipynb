>### Perceptrons:
- The fundamental building block of an Artificial Neural Network (ANN) is the Perceptron.
- A Perceptron typically consists of an input layer and an output layer.
- Perceptron functions very similarly to a Logistic Regression model.

**Formula:** Just like in linear/logistic regression where you calculate a weighted sum of inputs plus a bias ,the Perceptron performs the exact same calculation.
                    z = w1​.x1 ​+ w2​.x2​ + w3.​x3 ​+ b

**Activation Function:** After calculating the weighted sum, a Sigmoid Activation Function is applied to squash the output between 0 and 1. 
The formula used is:     **f(x) = 1/1+e^-1**

- This allows the model to perform binary classification (predicting, for example, 0 or 1, Yes or No).



**Limitation** 

**Linear Separability:** A single-layer perceptron can only learn linearly separable patterns. If the data cannot be divided by a straight line (a linear decision boundary), the perceptron fails to classify it correctly (e.g., the classic XOR problem).

**Key Concept:** while a basic Perceptron has no hidden layers, the addition of one or more hidden layers transforms the structure into a full-fledged Artificial Neural Network.

>### Backward Propagation
**Chain Rule of Calculus:** The network uses the chain rule to calculate how much each weight and bias contributed to the final error.

**Gradient Descent:** Once the gradients (partial derivatives of the loss with respect to each weight) are calculated, the model updates the weights.

**Weight Update Formula:**The update rule is expressed as: 


$$
w_{\text{new}} = w_{\text{old}} - \eta \frac{\partial L}{\partial w_{\text{old}}}
$$

>#### Gradient Descent

Gradient Descent is an optimization algorithm used to minimize the loss (error) of a neural network by iteratively updating the model's weights and biases. It calculates the gradient (partial derivative) of the loss function with respect to each parameter and adjusts the parameters in the opposite direction of the gradient. This process is repeated until the model converges to a minimum loss.

 **Working Steps**

1. Initialize weights and biases randomly.
2. Perform forward propagation to compute predictions.
3. Calculate the loss using a loss function.
4. Compute gradients using backpropagation.
5. Update weights and biases using Gradient Descent.
6. Repeat until the loss converges.


**Epoch:** one Epoch as the completion of one full forward pass and one full backward pass through the entire dataset.

**Training Iteration**: The model is trained over multiple epochs until the accuracy plateaus or the loss reaches a global minima (the lowest point of the loss function).

#### Types of Gradient Descent

1. **Batch Gradient Descent**
   - Uses the entire training dataset for one update.
   - Stable but computationally expensive.

2. **Stochastic Gradient Descent (SGD)**
   - Updates weights after each training example.
   - Faster but more noisy.

3. **Mini-Batch Gradient Descent**
   - Updates weights using a small batch of samples.
   - Most commonly used in deep learning because it balances speed and stability.

### Vanishing Gradient Problem

The **Vanishing Gradient Problem** occurs during backpropagation when the gradients become extremely small as they are propagated back through multiple hidden layers. As a result, the weights of the earlier layers are updated very little or not at all, causing the network to learn slowly or stop learning.

 **Why does it happen with Sigmoid?**

The Sigmoid activation function outputs values between **0 and 1**.

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

Its derivative is

$$
\sigma'(x)=\sigma(x)\left(1-\sigma(x)\right)
$$

The maximum value of the derivative is only **0.25**. During backpropagation, many such small derivatives are multiplied together, making the overall gradient approach **zero**.

 **Solution**

- Use **ReLU** or its variants instead of Sigmoid.
- Use proper weight initialization (He/Xavier Initialization).
- Apply Batch Normalization.
